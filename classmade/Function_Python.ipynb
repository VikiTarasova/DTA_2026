Типи результатів функцій

````python
my_dict = {
    1: "one",
    2: "two",
    3: "three",
    4: "four",
    5: "five"
}
````

Функція щось виконала, проте цей результат підставити ніде не можна


````python
# Повертая NoneType, без зміни даних, виводить результат
def info(d):
  print(d)

# info(my_dict)
type(info(my_dict))
````

````python
# Повертая NoneType, змінює початкові дані, виводить результат
def info(d):
  d[6] = "six"
  print(d)

# info(my_dict)
type(info(my_dict))
````

````python
# Повертая NoneType, змінює початкові дані, не виводить результат

def add_el(d):
  d[6] = "six"

print(type(add_el(my_dict)))
print(f"my_dict: {my_dict}")
````

Функція повертає результат для обробки

````python
def sum_num(a, b):
  print(a + b)

type(sum_num(5, 7))
````

````python
def sum_num(a, b):
  return a + b

print(sum_num(5, 7))
print(type(sum_num(5, 7)))

print()

sum_num(5, 7) + 10
````

````python
my_dict = {
    1: "one",
    2: "two",
    3: "three",
    4: "four",
    5: "five"
}
````

````python
def dict_value(d, key):
  print(d[key])

dict_value(my_dict, 3)
````

````python
def dict_value(d, key):
  return d[key]

dict_value(my_dict, 3) + "333"
````

````python
def dict_value(d, key):
  return d[key]

print(dict_value(my_dict, 3))
print(type(dict_value(my_dict, 3)))
dict_value(my_dict, 3) + "333"
````
