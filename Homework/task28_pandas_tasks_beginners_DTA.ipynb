## Підготовка: Генерація тестових даних

Запустіть цю комірку перед початком роботи. Не змінюйте код.

````python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 200

categories = ['Електроніка', 'Одяг', 'Книги', 'Продукти', 'Спорт']
cities = ['Київ', 'Харків', 'Одеса', 'Дніпро', 'Львів']
payment_methods = ['Картка', 'Готівка', 'Онлайн']

df = pd.DataFrame({
    'order_id':        range(1001, 1001 + n),
    'date':            pd.date_range('2023-01-01', periods=n, freq='D')[:n],
    'category':        np.random.choice(categories, n),
    'product':         ['Товар_' + str(i) for i in np.random.randint(1, 51, n)],
    'city':            np.random.choice(cities, n),
    'quantity':        np.random.randint(1, 10, n),
    'price':           np.round(np.random.uniform(50, 5000, n), 2),
    'payment_method':  np.random.choice(payment_methods, n),
    'rating':          np.random.choice([1, 2, 3, 4, 5, np.nan], n, p=[0.05, 0.1, 0.15, 0.3, 0.3, 0.1]),
    'is_returned':     np.random.choice([True, False], n, p=[0.1, 0.9]),
})

# Додаємо кілька пропущених значень
df.loc[np.random.choice(df.index, 10, replace=False), 'city'] = np.nan
df.loc[np.random.choice(df.index, 5, replace=False), 'price'] = np.nan

# Розраховуємо загальну суму
df['total'] = df['quantity'] * df['price']

print('Дані успішно згенеровано. Рядків:', len(df))
print('Колонки:', list(df.columns))
````

Відсортуйте результат за загальним обсягом продажів.

## Завдання 9 (Середнє). Додавання обчислюваних колонок

Додайте до датафрейму нові колонки:
1. `month` — місяць замовлення (з колонки `date`).
2. `price_category` — категорія ціни: `'Низька'` (до 500), `'Середня'` (500–2000), `'Висока'` (понад 2000). Використайте `pd.cut()` або `np.select()`.
3. `discount_total` — сума зі знижкою 15%.

````python
# Ваш код тут

# 1. Місяць замовлення
df['month'] = df['date'].dt.month

# 2. Категорія ціни
conditions = [
    df['price'] < 500,
    (df['price'] >= 500) & (df['price'] <= 2000),
    df['price'] > 2000
]

choices = ['Низька', 'Середня', 'Висока']

df['price_category'] = np.select(conditions, choices, default='Невідомо')

# 3. Сума зі знижкою 15%
df['discount_total'] = df['total'] * 0.85


print(df[['date', 'month', 'price', 'price_category', 'total', 'discount_total']].head())

````

````python

df['price_category'] = pd.cut(
    df['price'],
    bins=[0, 500, 2000, float('inf')],
    labels=['Низька', 'Середня', 'Висока']
)

print(df['price_category'].value_counts())
````

**Висновок:** *(Який розподіл цінових категорій? Як метод pd.cut спрощує категоризацію?)*  

Після категоризації всі товари були розподілені на три цінові сегменти: «Низька», «Середня» та «Висока». Це дозволяє аналізувати продажі не лише за конкретною ціною, а й за ціновими групами, що значно спрощує сегментацію клієнтів і товарів.  

Метод `pd.cut()` автоматично розподіляє числові значення по заданих інтервалах, тому не потрібно вручну перевіряти кожну ціну та присвоювати їй категорію. Завдяки цьому дані стають більш зручними для аналізу, побудови звітів і порівняння результатів між різними ціновими сегментами.

## Завдання 11 (Середнє). Підрахунок унікальних значень і частот

````python
# Ваш код тут
# 1. Унікальні товари у кожній категорії
unique_products = df.groupby('category')['product'].nunique()
print("Унікальні товари по категоріях:")
print(unique_products)

# 2. Топ-5 найпопулярніших товарів
top5 = df['product'].value_counts().head(5)
print("\nТоп-5 товарів:")
print(top5)

# 3. Частка способів оплати у відсотках
payment_share = df['payment_method'].value_counts(normalize=True) * 100
print("\nСпособи оплати (%):")
print(payment_share.round(2))
````

1. Додайте колонку `city_avg_total` — середня сума замовлення по місту (для кожного рядка підставте середнє по його місту), використайте `groupby().transform('mean')`.
2. Додайте колонку `above_city_avg` — булева колонка, що показує, чи перевищує замовлення середнє по місту.

````python
# Ваш код тут
# 1. Середня сума по місту для кожного рядка
df['city_avg_total'] = df.groupby('city')['total'].transform('mean')

# 2. Чи перевищує замовлення середнє по місту
df['above_city_avg'] = df['total'] > df['city_avg_total']

# 3. Частка замовлень вище середнього по кожному місту
city_above_avg = df.groupby('city')['above_city_avg'].mean() * 100

print(df[['city', 'total', 'city_avg_total', 'above_city_avg']].head(10))
print("\nЧастка замовлень вище середнього по містах (%):")
print(city_above_avg.round(2).sort_values(ascending=False))
````

**Висновок:** *(Чим transform відрізняється від звичайного groupby? У якому місті найбільша частка великих замовлень?)*  

Найбільша частка замовлень вище середнього по місту у Дніпрі — 42,11%. Це означає, що майже половина покупок у цьому місті є відносно великими.  

`transform()` відрізняється від `groupby()` тим, що:  

`groupby()` повертає агреговані значення (по групах),  
`transform()` повертає результат для кожного рядка, зберігаючи початковий розмір DataFrame (тобто “приклеює” групову статистику назад до кожного запису).

1. Згрупуйте дані за місяцем та розрахуйте щоденну виручку.

````python
# Ваш код тут
# 1. Щоденна виручка
daily_revenue = df.groupby('date')['total'].sum().reset_index()
daily_revenue.columns = ['date', 'revenue']
daily_revenue = daily_revenue.sort_values('date')

# 2. 7-денне ковзне середнє
daily_revenue['rolling_mean'] = daily_revenue['revenue'].rolling(7).mean()
daily_revenue['rolling_std']  = daily_revenue['revenue'].rolling(7).std()

# 3. Аномалії — відхилення більш ніж на 2 std
daily_revenue['is_anomaly'] = (
    (daily_revenue['revenue'] - daily_revenue['rolling_mean']).abs()
    > 2 * daily_revenue['rolling_std']
)

anomalies = daily_revenue[daily_revenue['is_anomaly']]

print(f"Знайдено аномалій: {len(anomalies)}")
print("\nДні-аномалії:")
print(anomalies[['date', 'revenue', 'rolling_mean', 'rolling_std']].round(2))
````

1. Для кожного товару (`product`) знайдіть місяць першого продажу — це його «когорта».
2. Об'єднайте цю інформацію з основним датафреймом.
3. Побудуйте зведену таблицю: когорта (місяць першого продажу) проти місяця замовлення, значення — кількість замовлень.

Підказка: використайте `groupby('product')['date'].transform('min')` та `dt.to_period('M')`.

````python
# Ваш код тут
# 1. Місяць першого продажу для кожного товару (когорта)
df['cohort'] = (
    df.groupby('product')['date']
    .transform('min')
    .dt.to_period('M')
)

# 2. Місяць поточного замовлення
df['order_month'] = df['date'].dt.to_period('M')

# 3. Зведена таблиця: когорта × місяць замовлення
cohort_table = df.pivot_table(
    index='cohort',
    columns='order_month',
    values='product',
    aggfunc='count'
).fillna(0).astype(int)

print("Коґортна таблиця (кількість замовлень):")
print(cohort_table)

print(f"\nРозмір таблиці: {cohort_table.shape}")
print(f"Когорт: {cohort_table.shape[0]}, Місяців: {cohort_table.shape[1]}")
````

## Завдання 20 (Складне). Підсумковий звіт

Сформуйте підсумковий звіт у вигляді одного датафрейму `summary_report`, що містить для кожної категорії:
- загальний дохід (`total_revenue`),
- кількість замовлень (`order_count`),
- середній чек (`avg_order_value`),
- частку доходу від загального (`revenue_share_pct`),
- рівень повернень у відсотках (`return_rate_pct`),
- середній рейтинг без NaN (`avg_rating`).

Відсортуйте за `total_revenue` спадно. Збережіть результат у файл `summary_report.csv`.

````python
# Ваш код тут
summary_report = (
    df.groupby('category')
    .agg(
        total_revenue   = ('total',       'sum'),
        order_count     = ('total',       'count'),
        avg_order_value = ('total',       'mean'),
        return_rate_pct = ('is_returned', 'mean'),
        avg_rating      = ('rating',      'mean'),
    )
    .reset_index()
)

# Частка доходу від загального
summary_report['revenue_share_pct'] = (
    summary_report['total_revenue'] / summary_report['total_revenue'].sum() * 100
)

# Повернення у відсотки та округлення
summary_report['return_rate_pct'] = summary_report['return_rate_pct'] * 100

summary_report = (
    summary_report
    .sort_values('total_revenue', ascending=False)
    .reset_index(drop=True)
    .round(2)
)

print(summary_report.to_string(index=False))

# Збереження у CSV
summary_report.to_csv('summary_report.csv', index=False, encoding='utf-8-sig')
print("\nФайл збережено: summary_report.csv")
````
