# Розділ 1. Словники (`dict`)

**Швидке нагадування:**

Словник — це колекція пар `ключ : значення`, створюється у фігурних дужках:
```python
student = {"name": "Олена", "age": 20, "city": "Київ"}
```

**Основні операції:**
- Доступ: `student["name"]`
- Додавання/зміна: `student["email"] = "olena@example.com"`
- Видалення: `del student["age"]`
- Перевірка: `if "name" in student:`
- Безпечне отримання: `student.get("phone")` — повертає `None`, якщо ключа немає
- Довжина: `len(student)` — кількість пар

**Перебір у циклі:**
- `for key in d.keys():` — перебір ключів
- `for value in d.values():` — перебір значень
- `for key, value in d.items():` — перебір пар

### Завдання 16. Підрахунок слів у тексті ⭐⭐

Дано словник з цінами. Виведіть товари відсортованими за ціною (від дешевого до дорогого) у форматі `Назва — Ціна грн`.

> 💡 **Підказка:** Найпростіший спосіб: створіть список пар через `list(prices.items())`. Потім у циклі багато разів знаходьте пару з мінімальною ціною, виводьте її і видаляйте зі списку. Або скористайтесь `sorted(prices.items(), key=lambda x: x[1])` — це коротший варіант, якщо ви вже бачили `lambda`.

````python
prices = {
    "хліб": 25,
    "молоко": 38,
    "сир": 120,
    "масло": 95,
    "мʼясо": 280,
    "риба": 250,
    "яйця": 65
}
# Ваш код:
пари = list(prices.items())

while пари:
    мін_пара = пари[0]
    for пара in пари:
        if пара[1] < мін_пара[1]:
            мін_пара = пара
    print(f"{мін_пара[0]} — {мін_пара[1]} грн")
    пари.remove(мін_пара)
````

### Завдання 20. Інвентар магазину ⭐⭐⭐

Дано словник з товарами магазину, де значення — це інший словник з ціною і кількістю на складі. Виконайте:
1. Виведіть загальну вартість всього товару на складі (ціна × кількість для кожного).
2. Виведіть назви товарів, яких залишилось менше 20 штук.
3. Виведіть найдорожчий товар.

````python
inventory = {
    "Ноутбук":   {"price": 18500, "stock": 12},
    "Телефон":   {"price": 9200,  "stock": 35},
    "Планшет":   {"price": 7800,  "stock": 8},
    "Навушники": {"price": 850,   "stock": 120},
    "Клавіатура":{"price": 1200,  "stock": 45},
    "Мишка":     {"price": 450,   "stock": 18}
}
# 1. Загальна вартість складу
загальна_вартість = 0
for товар, дані in inventory.items():
    загальна_вартість += дані["price"] * дані["stock"]

print(f"Загальна вартість складу: {загальна_вартість} грн")

# 2. Товари, яких менше 20 штук
print("\nТоварів менше 20 штук:")
for товар, дані in inventory.items():
    if дані["stock"] < 20:
        print(f"  {товар} — {дані['stock']} шт.")

# 3. Найдорожчий товар
макс_товар = None
макс_ціна = 0
for товар, дані in inventory.items():
    if дані["price"] > макс_ціна:
        макс_ціна = дані["price"]
        макс_товар = товар

print(f"\nНайдорожчий товар: {макс_товар} — {макс_ціна} грн")
````

### Завдання 21. Аналіз продажів по категоріях ⭐⭐⭐

Дано список покупок, де кожна покупка — це словник з полями `product`, `category`, `price`. Створіть новий словник, де ключ — категорія, а значення — загальна виручка по цій категорії.

> 💡 **Підказка:** Створіть порожній словник `revenue = {}`. У циклі для кожної покупки: якщо її категорія вже є у словнику — додайте ціну до існуючого значення, якщо немає — створіть запис з ціною як початковим значенням.

````python
purchases = [
    {"product": "Хліб",       "category": "Продукти",  "price": 25},
    {"product": "Молоко",     "category": "Продукти",  "price": 38},
    {"product": "Шампунь",    "category": "Гігієна",   "price": 120},
    {"product": "Зубна паста","category": "Гігієна",   "price": 65},
    {"product": "Сир",        "category": "Продукти",  "price": 180},
    {"product": "Книга",      "category": "Розваги",   "price": 220},
    {"product": "Мило",       "category": "Гігієна",   "price": 35},
    {"product": "Фільм",      "category": "Розваги",   "price": 150},
    {"product": "Ковбаса",    "category": "Продукти",  "price": 195}
]
# Ваш код:
revenue = {}

for покупка in purchases:
    категорія = покупка["category"]
    ціна = покупка["price"]

    if категорія in revenue:
        revenue[категорія] += ціна
    else:
        revenue[категорія] = ціна

for категорія, виручка in revenue.items():
    print(f"{категорія}: {виручка} грн")
````

### Завдання 22. Телефонна книга з пошуком ⭐⭐⭐

Створіть словник-телефонну книгу з 5 контактами. Напишіть програму, яка у циклі `while`:
1. Запитує у користувача дію: `пошук`, `додати`, `видалити`, `вихід`
2. Залежно від вибору — виконує відповідну дію
3. Завершується, коли користувач введе `вихід`

> 💡 **Підказка:** Використайте `while True:` (нескінченний цикл) і умову завершення з `break`, АБО змінну-прапорець, наприклад `running = True`.

````python
# Ваш код для завдання 22:
while True:
    дія = input("\nДія (пошук / додати / видалити / вихід): ").lower().strip()

    if дія == "пошук":
        name = input("Введіть ім'я: ").strip()
        if name in phonebook:
            print(f"Телефон {name}: {phonebook[name]}")
        else:
            print("Контакт не знайдений")

    elif дія == "додати":
        name = input("Введіть ім'я: ").strip()
        phone = input("Введіть телефон: ").strip()
        phonebook[name] = phone
        print(f"Контакт {name} додано")

    elif дія == "видалити":
        name = input("Введіть ім'я: ").strip()
        if name in phonebook:
            del phonebook[name]
            print(f"Контакт {name} видалено")
        else:
            print("Контакт не знайдений")

    elif дія == "вихід":
        print("До побачення!")
        break

    else:
        print("Невідома дія. Спробуйте ще раз")
````

Виведіть:

````python
attendance = {
    "Олена":   [1, 1, 1, 1, 1],
    "Андрій":  [1, 0, 1, 1, 0],
    "Марія":   [1, 1, 1, 1, 1],
    "Богдан":  [0, 1, 0, 1, 1],
    "Софія":   [1, 1, 1, 0, 1],
    "Петро":   [0, 0, 1, 1, 1],
    "Дарʼя":   [1, 1, 1, 1, 0],
    "Михайло": [1, 0, 0, 1, 1]
}
# Ваш код:
days = ["Понеділок", "Вівторок", "Середа", "Четвер", "П'ятниця"]

# 1. Скільки днів кожен студент був присутній
print("1. Кількість присутніх днів:")
for name, days_list in attendance.items():
    print(f"   {name}: {sum(days_list)} дні(в)")

# 2. Хто був присутній усі 5 днів
print("\n2. Присутні всі 5 днів:")
for name, days_list in attendance.items():
    if sum(days_list) == 5:
        print(f"   {name}")

# 3. У який день найбільше прогулів
absences = [0, 0, 0, 0, 0]
for days_list in attendance.values():
    for i in range(5):
        if days_list[i] == 0:
            absences[i] += 1

worst_day = absences.index(max(absences))
print(f"\n3. День з найбільшою кількістю прогулів: {days[worst_day]} ({max(absences)} прогули(ів))")

# 4. Загальний відсоток відвідуваності
total = sum(sum(d) for d in attendance.values())
total_possible = len(attendance) * 5
percentage = total / total_possible * 100
print(f"\n4. Загальний відсоток відвідуваності: {percentage:.1f}%")
````

### Завдання 26. Калькулятор з історією ⭐⭐

У циклі `while` запитуйте у користувача два числа і операцію (`+`, `-`, `*`, `/`). Обчислюйте результат і зберігайте кожен у списку `history`. Цикл завершується, коли користувач введе `вихід` замість операції. У кінці виведіть усю історію обчислень.

````python
# Ваш код для завдання 26:
history = []

while True:
    operation = input("Введіть операцію (+, -, *, /) або 'вихід': ")

    if operation == "вихід":
        break

    if operation not in ["+", "-", "*", "/"]:
        print("Невідома операція. Спробуйте ще раз.")
        continue

    a = float(input("Введіть перше число: "))
    b = float(input("Введіть друге число: "))

    if operation == "+":
        result = a + b
    elif operation == "-":
        result = a - b
    elif operation == "*":
        result = a * b
    elif operation == "/":
        if b == 0:
            print("Помилка: ділення на нуль!")
            continue
        result = a / b

    record = f"{a} {operation} {b} = {result}"
    history.append(record)
    print(f"Результат: {result}")

print("\n Історія обчислень:")
if history:
    for i, record in enumerate(history, 1):
        print(f"   {i}. {record}")
else:
    print("   Історія порожня.")
````

4. Коли користувач введе `завершити` — виведіть усі товари в корзині та загальну суму

3. Кількість студентів, які мають середній бал вище загального
4. Студента з найвищим середнім балом

````python
grades = {
    "Олена":   [85, 90, 78, 92, 88],
    "Андрій":  [72, 65, 80, 75, 70],
    "Марія":   [95, 92, 88, 90, 94],
    "Богдан":  [60, 65, 70, 55, 68],
    "Софія":   [82, 85, 80, 88, 84],
    "Петро":   [78, 75, 82, 80, 76],
    "Дарʼя":   [90, 88, 92, 85, 91]
}
# Ваш код:
# 1. Середній бал кожного студента
averages = {}
print("1. Середній бал кожного студента:")
for name, marks in grades.items():
    avg = sum(marks) / len(marks)
    averages[name] = avg
    print(f"   {name}: {avg:.1f}")

# 2. Загальний середній бал групи
group_avg = sum(averages.values()) / len(averages)
print(f"\n2. Загальний середній бал групи: {group_avg:.1f}")

# 3. Студенти вище загального середнього
above_avg = [name for name, avg in averages.items() if avg > group_avg]
print(f"\n3. Студентів з балом вище загального ({group_avg:.1f}): {len(above_avg)}")
for name in above_avg:
    print(f"   - {name}: {averages[name]:.1f}")

# 4. Студент з найвищим середнім балом
best = max(averages, key=averages.get)
print(f"\n4. Найкращий студент: {best} ({averages[best]:.1f})")
````

### Завдання 35. Рейтинг команд ⭐⭐

Дано список матчів. Кожен матч — це словник `{"team_a": ..., "team_b": ..., "score_a": ..., "score_b": ...}`. Створіть таблицю очок: за перемогу — 3 очка, за нічию — 1, за поразку — 0. Виведіть команди у порядку від найбільшої до найменшої кількості очок.

> 💡 **Підказка:** Створіть словник `points = {}`. У циклі по матчах: визначайте переможця і додавайте очки. Не забудьте обробити випадок, коли команди ще немає у словнику.

````python
matches = [
    {"team_a": "Динамо",     "team_b": "Шахтар",     "score_a": 1, "score_b": 2},
    {"team_a": "Зоря",       "team_b": "Динамо",     "score_a": 0, "score_b": 1},
    {"team_a": "Шахтар",     "team_b": "Дніпро",     "score_a": 3, "score_b": 1},
    {"team_a": "Динамо",     "team_b": "Ворскла",    "score_a": 2, "score_b": 2},
    {"team_a": "Шахтар",     "team_b": "Зоря",       "score_a": 4, "score_b": 0},
    {"team_a": "Дніпро",     "team_b": "Ворскла",    "score_a": 1, "score_b": 1},
    {"team_a": "Зоря",       "team_b": "Дніпро",     "score_a": 2, "score_b": 1},
    {"team_a": "Ворскла",    "team_b": "Шахтар",     "score_a": 0, "score_b": 3}
]
# Ваш код:

points = {}

for match in matches:
    team_a  = match["team_a"]
    team_b  = match["team_b"]
    score_a = match["score_a"]
    score_b = match["score_b"]

    # Додаємо команди у словник якщо ще немає
    if team_a not in points:
        points[team_a] = 0
    if team_b not in points:
        points[team_b] = 0

    # Нараховуємо очки
    if score_a > score_b:       # перемога team_a
        points[team_a] += 3
    elif score_b > score_a:     # перемога team_b
        points[team_b] += 3
    else:                       # нічия
        points[team_a] += 1
        points[team_b] += 1

# Сортуємо від більшого до меншого
ranking = sorted(points.items(), key=lambda x: x[1], reverse=True)

print("Рейтинг команд:\n")
print(f"{'Місце':<7} {'Команда':<12} {'Очки'}")
print("-" * 28)
for place, (team, pts) in enumerate(ranking, 1):
    print(f"{place:<7} {team:<12} {pts}")
````

````python
import random

wins = {"користувач": 0, "комп'ютер": 0}
options = ["камінь", "ножиці", "папір"]

beats = {
    "камінь":  "ножиці",
    "ножиці":  "папір",
    "папір":   "камінь"
}

print("Гра 'Камінь-Ножиці-Папір'")
print("Перший до 3 перемог виграє!\n")

while wins["користувач"] < 3 and wins["комп'ютер"] < 3:
    u = wins["користувач"]
    c = wins["комп'ютер"]
    round_num = u + c + 1
    print(f"--- Раунд {round_num} | Рахунок: Ви {u} : {c} Комп'ютер ---")

    user = input("Ваш вибір (камінь / ножиці / папір): ").strip().lower()

    if user not in options:
        print("Невірний вибір, спробуйте ще раз.\n")
        continue

    computer = random.choice(options)
    print(f"Комп'ютер обрав: {computer}")

    if user == computer:
        print("Нічия!\n")
    elif beats[user] == computer:
        wins["користувач"] += 1
        print(f"Ви перемогли! ({user} б'є {computer})\n")
    else:
        wins["комп'ютер"] += 1
        print(f"Комп'ютер переміг! ({computer} б'є {user})\n")

print("=" * 40)
u = wins["користувач"]
c = wins["комп'ютер"]
if u == 3:
    print(f"ВИ ВИГРАЛИ з рахунком {u} : {c}!")
else:
    print(f" Комп'ютер виграв з рахунком {c} : {u}!")
````

Створіть словник, де кожній букві українського алфавіту відповідає інша буква (зсув на 1: `а` → `б`, `б` → `в`, ... `я` → `а`). Зашифруйте слово, введене користувачем, використовуючи цей словник.

> 💡 **Підказка:** Спочатку створіть словник у циклі. Алфавіт: `'абвгґдеєжзиіїйклмнопрстуфхцчшщьюя'`. Для кожної літери з номером `i` ставте літеру з номером `(i + 1) % len(alphabet)`.

````python
# Ваш код для завдання 44:
# Український алфавіт
alphabet = "абвгґдеєжзиіїйклмнопрстуфхцчшщьюя"

# Створення словника шифрування
cipher = {}

for i in range(len(alphabet)):
    current_letter = alphabet[i]
    next_letter = alphabet[(i + 1) % len(alphabet)]

    cipher[current_letter] = next_letter

# Введення слова
word = input("Введіть слово: ").lower()

# Шифрування
encrypted_word = ""

for letter in word:

    # Якщо літера є в алфавіті
    if letter in cipher:
        encrypted_word += cipher[letter]

    # Якщо символ не літера — залишаємо без змін
    else:
        encrypted_word += letter

# Вивід результату
print("Зашифроване слово:", encrypted_word)
````

- `вихід` — завершити

Зберігайте кожну операцію у списку як словник `{"type": "зняти", "amount": 500}`.

> 💡 **Підказка:** Створіть `history = []` перед циклом. Кожну операцію додавайте через `.append()`.

````python
# Ваш код для завдання 45:
balance = 5000
history = []

while True:
    action = input("\nДія (баланс / зняти / поповнити / історія / вихід): ").strip().lower()

    if action == "баланс":
        print(f"Поточний баланс: {balance} грн")

    elif action == "зняти":
        amount = int(input("Сума для зняття: "))
        if amount <= 0:
            print("Сума повинна бути більше нуля.")
        elif amount > balance:
            print(f"Недостатньо коштів. Доступно: {balance} грн")
        else:
            balance -= amount
            history.append({"type": "зняти", "amount": amount})
            print(f"Знято {amount} грн. Залишок: {balance} грн")

    elif action == "поповнити":
        amount = int(input("Сума для поповнення: "))
        if amount <= 0:
            print("Сума повинна бути більше нуля.")
        else:
            balance += amount
            history.append({"type": "поповнити", "amount": amount})
            print(f"Поповнено на {amount} грн. Баланс: {balance} грн")

    elif action == "історія":
        if not history:
            print("Операцій ще немає.")
        else:
            print("\n--- Історія операцій ---")
            for op in history:
                sign = "+" if op["type"] == "поповнити" else "-"
                print(f"  {op['type']:>10}: {sign}{op['amount']} грн")

    elif action == "вихід":
        print(f"До побачення! Ваш баланс: {balance} грн")
        break

    else:
        print("Невідома команда. Спробуйте ще раз.")
````

### Завдання 46. Менеджер задач (todo) ⭐⭐⭐

Створіть програму-менеджер задач. Задачі зберігаються у списку словників: `{"task": "...", "done": False}`. У циклі `while` запитуйте дію:
- `додати` — додати нову задачу
- `показати` — показати всі задачі з номерами і статусом
- `виконати` — за номером позначити задачу як виконану
- `видалити` — за номером видалити задачу
- `вихід` — завершити

````python
# Ваш код для завдання 46:
tasks = []

while True:
    action = input("\nДія (додати / показати / виконати / видалити / вихід): ").strip().lower()

    if action == "додати":
        text = input("Назва задачі: ").strip()
        if text:
            tasks.append({"task": text, "done": False})
            print(f"Задачу «{text}» додано.")
        else:
            print("Задача не може бути порожньою.")

    elif action == "показати":
        if not tasks:
            print("Список задач порожній.")
        else:
            print("\n--- Список задач ---")
            for i, t in enumerate(tasks, start=1):
                status = "[x]" if t["done"] else "[ ]"
                print(f"  {i}. {status} {t['task']}")

    elif action == "виконати":
        if not tasks:
            print("Немає задач.")
        else:
            num = int(input(f"Номер задачі (1–{len(tasks)}): "))
            if 1 <= num <= len(tasks):
                tasks[num - 1]["done"] = True
                print(f"Задачу «{tasks[num - 1]['task']}» виконано.")
            else:
                print("Невірний номер.")

    elif action == "видалити":
        if not tasks:
            print("Немає задач.")
        else:
            num = int(input(f"Номер задачі (1–{len(tasks)}): "))
            if 1 <= num <= len(tasks):
                removed = tasks.pop(num - 1)
                print(f"Задачу «{removed['task']}» видалено.")
            else:
                print("Невірний номер.")

    elif action == "вихід":
        done = sum(1 for t in tasks if t["done"])
        print(f"До побачення! Виконано {done} з {len(tasks)} задач.")
        break

    else:
        print("Невідома команда. Спробуйте ще раз.")
````

### Завдання 48. Аналіз продажів магазину ⭐⭐⭐

Дано список покупок у форматі словників `{"customer": "...", "product": "...", "price": ..., "day": ...}`. Створіть звіт:
1. Загальна виручка
2. Виручка по днях (словник `{day: revenue}`)
3. Найкращий клієнт (хто витратив найбільше)
4. Найпопулярніший товар (за кількістю покупок)
5. Середній чек (загальна виручка / кількість покупок)

````python
sales = [
    {"customer": "Олена",   "product": "Кава",     "price": 65, "day": 1},
    {"customer": "Андрій",  "product": "Чай",      "price": 45, "day": 1},
    {"customer": "Марія",   "product": "Кава",     "price": 65, "day": 1},
    {"customer": "Олена",   "product": "Тістечко", "price": 95, "day": 2},
    {"customer": "Богдан",  "product": "Кава",     "price": 65, "day": 2},
    {"customer": "Софія",   "product": "Чай",      "price": 45, "day": 2},
    {"customer": "Петро",   "product": "Кава",     "price": 65, "day": 3},
    {"customer": "Дарʼя",   "product": "Тістечко", "price": 95, "day": 3},
    {"customer": "Михайло", "product": "Кава",     "price": 65, "day": 3},
    {"customer": "Олена",   "product": "Кава",     "price": 65, "day": 3},
    {"customer": "Андрій",  "product": "Тістечко", "price": 95, "day": 4},
    {"customer": "Марія",   "product": "Чай",      "price": 45, "day": 4},
    {"customer": "Олена",   "product": "Тістечко", "price": 95, "day": 5}
]
# Ваш код:


# 1. Загальна виручка
total_revenue = 0

# 2. Виручка по днях
revenue_by_day = {}

# 3. Витрати клієнтів
customer_spending = {}

# 4. Кількість покупок товарів
product_count = {}

for sale in sales:
    customer = sale["customer"]
    product = sale["product"]
    price = sale["price"]
    day = sale["day"]

    # Загальна виручка
    total_revenue += price

    # Виручка по днях
    if day not in revenue_by_day:
        revenue_by_day[day] = 0
    revenue_by_day[day] += price

    # Витрати клієнтів
    if customer not in customer_spending:
        customer_spending[customer] = 0
    customer_spending[customer] += price

    # Популярність товарів
    if product not in product_count:
        product_count[product] = 0
    product_count[product] += 1

# Найкращий клієнт
best_customer = max(customer_spending, key=customer_spending.get)

# Найпопулярніший товар
most_popular_product = max(product_count, key=product_count.get)

# Середній чек
average_check = total_revenue / len(sales)

# Вивід результатів
print("Загальна виручка:", total_revenue)
print("Виручка по днях:", revenue_by_day)
print("Найкращий клієнт:", best_customer)
print("Найпопулярніший товар:", most_popular_product)
print("Середній чек:", round(average_check, 2))
````

Реалізуйте гру: дано стартове слово. Гравець має ввести наступне слово, яке починається на ОСТАННЮ літеру попереднього. Якщо слово вже використовувалось — програш. Якщо слово починається не на ту літеру — програш. Цикл триває, поки гравець не програє або не введе `здаюсь`.

> 💡 **Підказка:** Зберігайте всі використані слова у списку. Першу і останню літеру отримуйте через індекси: `word[0]` і `word[-1]`.

````python
# Ваш код для завдання 52:
# Стартове слово
current_word = "кіт"

# Список використаних слів
used_words = [current_word]

print("Гра в слова!")
print("Стартове слово:", current_word)
print("Для виходу введіть: здаюсь")

while True:
    # Введення нового слова
    new_word = input("Введіть слово: ").lower()

    # Вихід із гри
    if new_word == "здаюсь":
        print("Гру завершено!")
        break

    # Перевірка: слово вже використовувалось
    if new_word in used_words:
        print("Програш! Це слово вже було використано.")
        break

    # Перевірка першої літери
    if new_word[0] != current_word[-1]:
        print("Програш! Слово повинно починатися на літеру:", current_word[-1])
        break

    # Додаємо слово у список
    used_words.append(new_word)

    # Оновлюємо поточне слово
    current_word = new_word

    print("Добре! Наступна літера:", current_word[-1])

# Вивід усіх використаних слів
print("Використані слова:", used_words)
````

### Завдання 53. Аналіз журналу ⭐⭐⭐

Дано список словників-записів журналу логів. Кожен запис: `{"level": "INFO"|"WARNING"|"ERROR", "message": "...", "hour": ...}`. Створіть звіт:
1. Скільки записів кожного рівня (словник)
2. У яку годину було найбільше ERROR-ів
3. Усі повідомлення з рівнем ERROR
4. Чи були години без жодного ERROR (з тих, у яких є записи)

````python
logs = [
    {"level": "INFO",    "message": "Сервер запущено",     "hour": 8},
    {"level": "INFO",    "message": "Користувач увійшов", "hour": 9},
    {"level": "WARNING", "message": "Повільний запит",     "hour": 9},
    {"level": "ERROR",   "message": "Помилка БД",          "hour": 10},
    {"level": "ERROR",   "message": "Тайм-аут запиту",     "hour": 10},
    {"level": "INFO",    "message": "Запит виконано",      "hour": 11},
    {"level": "WARNING", "message": "Низько місця на диску","hour": 12},
    {"level": "ERROR",   "message": "Помилка БД",          "hour": 14},
    {"level": "INFO",    "message": "Резервна копія",      "hour": 15},
    {"level": "ERROR",   "message": "Збій авторизації",    "hour": 14},
    {"level": "ERROR",   "message": "Помилка БД",          "hour": 14},
    {"level": "INFO",    "message": "Користувач вийшов",   "hour": 17}
]
# Ваш код:
# 1. Кількість записів кожного рівня
level_count = {}

# ERROR-и по годинах
error_by_hour = {}

# 3. Повідомлення ERROR
error_messages = []

# Усі години з записами
all_hours = []

for log in logs:
    level = log["level"]
    message = log["message"]
    hour = log["hour"]

    # Підрахунок рівнів
    if level not in level_count:
        level_count[level] = 0

    level_count[level] += 1

    # Зберігаємо години
    if hour not in all_hours:
        all_hours.append(hour)

    # Аналіз ERROR
    if level == "ERROR":

        # Повідомлення ERROR
        error_messages.append(message)

        # ERROR-и по годинах
        if hour not in error_by_hour:
            error_by_hour[hour] = 0

        error_by_hour[hour] += 1

# 2. Година з найбільшою кількістю ERROR
max_error_hour = max(error_by_hour, key=error_by_hour.get)

# 4. Години без ERROR
hours_without_error = []

for hour in all_hours:
    if hour not in error_by_hour:
        hours_without_error.append(hour)

# Вивід результатів
print("Кількість записів кожного рівня:")
print(level_count)

print("\nГодина з найбільшою кількістю ERROR:")
print(max_error_hour)

print("\nПовідомлення ERROR:")
for message in error_messages:
    print("-", message)

print("\nГодини без ERROR:")
print(hours_without_error)
````
